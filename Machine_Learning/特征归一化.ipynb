### 最大最小值归一化

```python
import numpy as np
from sklearn import datasets
```

```python
iris = datasets.load_iris()    # 加载鸢尾花数据集
X = iris.data
y = iris.target
```

```python
X[:,0] = (X[:,0] - np.min(X[:,0])) / (np.max(X[:,0]) - np.min(X[:,0]))
X[:,1] = (X[:,1] - np.min(X[:,1])) / (np.max(X[:,1]) - np.min(X[:,1]))
X[:,2] = (X[:,2] - np.min(X[:,2])) / (np.max(X[:,2]) - np.min(X[:,2]))
X[:,3] = (X[:,3] - np.min(X[:,3])) / (np.max(X[:,3]) - np.min(X[:,3]))
```

### scikit-learn 中的 StandardScaler

```python
from sklearn.preprocessing import StandardScaler
```

```python
iris = datasets.load_iris()
X = iris.data
y = iris.target
```

```python
X[:5]
```

```python
standard_scaler = StandardScaler()    # 创建 StandardScaler 标准化器
standard_scaler.fit(X);               # 根据 X 计算每个特征的均值和标准差
```

```python
standard_scaler.mean_    # 查看每个特征的均值
```

```python
standard_scaler.scale_   # 查看每个特征的标准差
```

```python
X = standard_scaler.transform(X)    # 使用计算出的均值和标准差对 X 进行标准化
X[:5]
```
