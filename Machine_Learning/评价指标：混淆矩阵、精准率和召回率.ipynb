```python
import numpy as np
from sklearn import datasets
```

```python
iris = datasets.load_iris()  # 加载鸢尾花数据集
X = iris.data
y = iris.target.copy()
y
```

```python
y[y != 0] = 1  # 转为二分类问题
y
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=666)
logistic_regression = LogisticRegression()
logistic_regression.fit(X_train, y_train)
y_predict = logistic_regression.predict(X_test)
```

### 手动计算 TN、FP、FN、TP、precision、recall、f1_score

```python
TN = np.sum((y_predict == 0) & (y_test == 0))
TN
```

```python
FP = np.sum((y_predict == 1) & (y_test == 0))
FP
```

```python
FN = np.sum((y_predict == 0) & (y_test == 1))
FN
```

```python
TP = np.sum((y_predict == 1) & (y_test == 1))
TP
```

```python
confusion_matrix = np.array([
    [TN, FP],
    [FN, TP]
])
confusion_matrix
```

```python
precision = TP / (FP + TP)
precision
```

```python
recall = TP / (FN + TP)
recall
```

```python
f1_score = 2 * precision * recall / (precision + recall)
f1_score
```

### sklearn中计算 TN、FP、FN、TP、precision、recall、f1_score

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test, y_predict)
```

```python
from sklearn.metrics import precision_score
precision_score(y_test, y_predict)
```

```python
from sklearn.metrics import recall_score
recall_score(y_test, y_predict)
```

```python
from sklearn.metrics import f1_score
f1_score(y_test, y_predict)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_predict)
```
