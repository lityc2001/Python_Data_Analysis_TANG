### 数据集

```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use("default")
```

```python
from sklearn.datasets import load_digits
digits = load_digits()
X = digits.data
y = digits.target
```

```python
X.shape, y.shape
```

```python
print(digits.DESCR)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=233)
```

### PCA降维

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA()
pca.fit(X_train);
```

```python
pca.explained_variance_         # 表示每个主成分解释的方差大小
```

```python
pca.explained_variance_ratio_   # 表示每个主成分解释了原始数据总方差的多少比例
```

```python
ratio_cum = np.cumsum(pca.explained_variance_ratio_)
ratio_cum
```

```python
plt.plot(ratio_cum)
plt.show()
```

```python
# 方法1
pca = PCA(20)  # 创建PCA模型，将数据降维到20个主成分
pca.fit(X_train);
```

```python
# 方法2
pca = PCA(0.9)  # 自动选择主成分数量，使累计解释方差比例至少达到90%
pca.fit(X_train);
pca.n_components_  # 查看实际保留的主成分数量
```

```python
X_train_pca = pca.transform(X_train)
X_test_pca = pca.transform(X_test)
```

```python
X_train.shape, X_train_pca.shape
```

### 降维的作用-模型训练

```python
from sklearn.linear_model import LogisticRegression
clf = LogisticRegression(
    solver='saga',
    tol=0.0001,
    max_iter=500,
    random_state=233)
```

```python
%%time
clf.fit(X_train, y_train)
clf.score(X_test, y_test)
```

```python
%%time
clf.fit(X_train_pca, y_train)
clf.score(X_test_pca, y_test)
```

### 降维的作用-可视化

```python
pca = PCA(2)  # 创建 PCA 模型，将原始数据降维到 2 个主成分
pca.fit(X_train);
```

```python
X_pca_2d = pca.transform(X_test)  # 将测试集转换到二维主成分空间
X_pca_2d.shape
```

```python
plt.rcParams["figure.figsize"] = (12, 8)
for i, digit in enumerate(y_test):
    plt.scatter(
        X_pca_2d[i, 0],
        X_pca_2d[i, 1],
        color=plt.cm.Dark2(digit),
        marker="${0}$".format(digit),
        s=60,
        alpha=0.5)
plt.show()
```
