### 拆分数据集

```python
shuffle_index = np.random.permutation(len(y))  # 随机生成0～149的不重复索引，用于打乱数据顺序
shuffle_index  # 查看打乱后的索引
```

```python
train_ratio = 0.8  # 设置训练集比例为80%
```

```python
train_size = int(len(y) * train_ratio)  # 计算训练集样本数量：150 × 0.8 = 120
train_size  # 查看训练集样本数量
```

```python
train_index = shuffle_index[:train_size]  # 前120个索引作为训练集索引
test_index = shuffle_index[train_size:]   # 剩余30个索引作为测试集索引
```

```python
X_train = X[train_index]   # 根据训练集索引提取特征数据
y_train = y[train_index]   # 根据训练集索引提取目标标签
X_test = X[test_index]     # 根据测试集索引提取特征数据
y_test = y[test_index]     # 根据测试集索引提取目标标签
```

```python
print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

```python
# 使用 train_test_split 分层拆分训练集与测试集
from sklearn.model_selection import train_test_split  # 导入数据集拆分函数
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, random_state=666, shuffle=True, stratify=y)  # 按 8:2 的比例拆分训练集和测试集
```

```python
print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

### 预测

```python
from sklearn.neighbors import KNeighborsClassifier  # 导入 KNN 分类模型
knn_classifier = KNeighborsClassifier(n_neighbors=5)  # 创建 KNN 分类器，设置邻居数量为 5
```

```python
knn_classifier.fit(X_train, y_train);  # 使用训练集训练模型
knn_classifier.score(X_test, y_test)  # 计算模型在测试集上的准确率
```

```python
y_predict = knn_classifier.predict(X_test)  # 使用训练好的模型预测测试集类别
y_predict  # 查看预测结果
```

### 评价

```python
accuracy = np.sum(y_predict == y_test)/len(y_test)  # 手动计算预测准确率
accuracy  # 查看准确率
```

```python
from sklearn.metrics import accuracy_score  # 导入准确率评价函数
accuracy_score(y_test, y_predict)  # 使用 accuracy_score 计算预测准确率
```
