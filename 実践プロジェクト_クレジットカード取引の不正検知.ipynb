### 1. 加载数据集

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = pd.read_csv("creditcard.csv")  # 训练集
```

```python
data[10000:10010]
```

### 2. 特征归一化

```python
from sklearn.preprocessing import StandardScaler  # 导入标准化工具 StandardScaler
data['V29'] = StandardScaler().fit_transform(data['Amount'].values.reshape(-1, 1))  # 对 Amount 交易金额进行标准化，并保存为新的特征 V29
data = data.drop(['Time', 'Amount'], axis=1)  # 删除原始的 Time 和 Amount 特征
data[:10]  # 查看处理后的前 10 行数据
```

### 3. 解决样本不均衡

```python
# 分离特征 X 和目标变量 y
X = data.loc[:,data.columns != 'Class']
y = data.loc[:,data.columns == 'Class']
```

```python
c0 = y['Class'].value_counts(sort=True).sort_index()  # 统计 Class 中各类别的样本数量，并按类别标签排序
c0
```

```python
# 使用下采样解决不均衡问题
n = len(y[y.Class == 1])  # 获取欺诈交易（Class=1）的样本数量
i0 = np.array(data[data.Class == 0].index)  # 获取正常交易（Class=0）的索引
i1 = np.array(data[data.Class == 1].index)  # 获取欺诈交易（Class=1）的索引
i0r = np.random.choice(i0, n, replace=False)  # 从正常交易中随机抽取 n 个样本，使数量与欺诈交易一致
i0r = np.array(i0r)

under_sample_ind = np.concatenate([i1, i0r])  # 合并全部欺诈样本索引和随机抽取的正常样本索引
under_sample_data = data.iloc[under_sample_ind,:]  # 根据索引生成下采样后的数据集

# 分离下采样数据中的特征和目标变量
X_undersample = under_sample_data.loc[:, under_sample_data.columns != 'Class']
y_undersample = under_sample_data.loc[:, under_sample_data.columns == 'Class']
```

### 4. 建模评估

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score
import warnings
warnings.filterwarnings('ignore')

# 定义模型比较函数
def model_selection(x, y, cv):
    # 创建不同的分类模型
    knn = KNeighborsClassifier(n_neighbors=3)  # K近邻，K=3
    lr = LogisticRegression()
    svc = SVC()
    gaussian = GaussianNB()
    dt = DecisionTreeClassifier(max_depth=5)
    rf = RandomForestClassifier(n_estimators=100)

    # 使用交叉验证评估各模型，并计算平均得分
    score_knn = np.mean(cross_val_score(knn, x, y, cv = cv))
    score_log = np.mean(cross_val_score(lr, x, y, cv = cv))
    score_svc = np.mean(cross_val_score(svc, x, y, cv = cv))
    score_gaussian = np.mean(cross_val_score(gaussian, x, y, cv = cv))
    score_dt_clf = np.mean(cross_val_score(dt, x, y, cv = cv))
    score_rf_clf = np.mean(cross_val_score(rf, x, y, cv = cv))

    # 将模型名称和对应的平均得分整理成 DataFrame
    models = pd.DataFrame({
    'Model': [ 'KNN', 'Logistic Regression', 'SVM',
              'Naive Bayes', 'Decision Tree','Random Forest' ],
    'Score': [score_knn, score_log, score_svc,
              score_gaussian, score_dt_clf, score_rf_clf]})
    # 按模型得分从高到低排序并输出
    print(models.sort_values(by='Score', ascending=False))
```

```python
cv = 5  # 设置为5折交叉验证
model_selection(X_undersample, y_undersample, cv)  # 使用下采样后的数据比较不同分类模型
```

### 5. 参数选择

```python
# 设置逻辑回归需要搜索的超参数范围
params = {
    'C': [0.1, 0.5, 1, 10, 100],            # 正则化强度的倒数：C 越小，正则化越强
    'penalty': ['l1', 'l2', 'elasticnet']   # 正则化类型
}
```

```python
from sklearn.model_selection import GridSearchCV  # 导入网格搜索工具
# 创建 GridSearchCV，对逻辑回归的不同参数组合进行搜索
grid = GridSearchCV(
    estimator = LogisticRegression(),  # 使用逻辑回归模型
    param_grid=params,                 # 指定需要搜索的参数组合
    n_jobs=-1
)
```

```python
grid.fit(X_undersample, y_undersample);  # 使用下采样后的数据进行网格搜索和模型训练
```

```python
grid.best_params_  # 查看网格搜索得到的最佳参数组合
```

```python
grid.best_score_  # 查看最佳参数组合对应的交叉验证平均得分
```

### 6. 绘制PR曲线

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
X_train, X_test, y_train, y_test = train_test_split(X_undersample, y_undersample, random_state=666)
model = grid.best_estimator_  # 使用网格搜索得到的最佳逻辑回归模型
model.fit(X_train,y_train)    # 使用训练集训练模型
y_predict = model.predict(X_test)  # 对测试集进行类别预测
decision_scores = model.decision_function(X_test)  # 获取测试样本的决策分数，用于绘制 PR 曲线
```

```python
from sklearn.metrics import precision_recall_curve  # 导入 Precision-Recall 曲线计算函数
# 根据真实标签和决策分数，计算不同阈值下的 Precision 和 Recall
precision_scores, recall_scores,thresholds =  precision_recall_curve(y_test,decision_scores)
```

```python
plt.plot(thresholds, precision_scores[:-1], color='r',label="precision")  # 绘制不同分类阈值下 Precision 的变化
plt.plot(thresholds, recall_scores[:-1], color='b',label="recall")        # 绘制不同分类阈值下 Recall 的变化
plt.legend()
plt.show()
```

```python
plt.plot(recall_scores,precision_scores)  # 绘制 Precision-Recall 曲线
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.show()
```

```python
from sklearn.metrics import roc_auc_score  # 导入 ROC-AUC 指标
auc = roc_auc_score(y_test,decision_scores)  # 根据真实标签和决策分数计算 ROC-AUC
print(auc)
```

```python
# 导入 Average Precision，用于评价 PR 曲线整体性能
from sklearn.metrics import average_precision_score
# 计算 AUPRC（Average Precision）
auprc = average_precision_score(y_test, decision_scores)
print(auprc)
```
