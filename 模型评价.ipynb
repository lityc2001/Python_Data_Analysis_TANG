### 拆分数据集

```python
shuffle_index = np.random.permutation(len(y))
shuffle_index
```

```python
train_ratio = 0.8
```

```python
train_size = int(len(y) * train_ratio)
train_size
```

```python
train_index = shuffle_index[:train_size]
test_index = shuffle_index[train_size:]
```

```python
X_train = X[train_index]
y_train = y[train_index]

X_test = X[test_index]
y_test = y[test_index]
```

```python
print(X_train.shape)
print(y_train.shape)
```

```python
print(X_test.shape)
print(y_test.shape)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, random_state=666, shuffle=True, stratify=y)
```

```python
print(X_train.shape)
print(y_train.shape)
```

```python
print(X_test.shape)
print(y_test.shape)
```

### 预测

```python
from sklearn.neighbors import KNeighborsClassifier
knn_classifier = KNeighborsClassifier(n_neighbors=5)
```

```python
knn_classifier.fit(X_train, y_train);
```

```python
knn_classifier.score(X_test, y_test)
```

```python
y_predict = knn_classifier.predict(X_test)
y_predict
```

### 评价

```python
accuracy = np.sum(y_predict == y_test)/len(y_test)
accuracy
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_predict)
```
