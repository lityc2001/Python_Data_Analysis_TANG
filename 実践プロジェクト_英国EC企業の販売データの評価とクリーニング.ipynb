# 英国EC企業の販売データの評価とクリーニング

## 分析目標

このデータ分析の目的は、市場の販売データに基づいて売れ筋商品を掘り起こし、売上向上につながるより効果的なマーケティング戦略を策定することである。
本実践プロジェクトの目的は、データのクリーンさと整然性を評価する練習を行い、その評価結果に基づいてデータをクリーニングすることで、次の分析に利用できるデータを得ることである。

## プロジェクト概要

元のデータセットには、英国のオンライン小売企業における2010年12月1日から2011年12月9日までのすべての取引記録が含まれており、世界各国・地域における同社の事業データを網羅している。同社は主に、誕生日プレゼント、結婚記念品、クリスマスギフトなど、さまざまな場面に対応したギフト商品を販売している。顧客層は主に卸売業者と個人消費者で構成されており、その中でも卸売業者がかなり大きな割合を占めている。

データの各列の意味は以下のとおりである。

* `InvoiceNo`：請求書番号。6桁の数字で、取引の一意の識別子として使用される。このコードが文字「C」で始まる場合、その取引がキャンセルされたことを示す。
* `StockCode`：商品コード。5桁の数字で、商品の一意の識別子として使用される。
* `Description`：商品名。
* `Quantity`：取引における商品の数量。
* `InvoiceDate`：請求書の日付と時刻。取引が発生した日時を表す。
* `UnitPrice`：単価。価格の単位は英ポンド（£）である。
* `CustomerID`：顧客番号。5桁の数字で、顧客の一意の識別子として使用される。
* `Country`：国名。顧客が居住している国の名称を表す。


## データの読み込み

データ分析に必要なライブラリをインポートし、Pandasの`read_csv`関数を使用して、元データファイル`e_commerce.csv`のデータ内容をDataFrameとして読み込み、変数`original_data`に代入する。

```python
import pandas as pd
```

```python
original_data = pd.read_csv("e_commerce.csv")
```

```python
original_data.head()
```

## データの評価

この部分では、前の部分で作成した`original_data`というDataFrameに含まれるデータを評価する。

評価は主に、構造と内容という2つの観点から行う。つまり、データの整然性とクリーンさを確認する。データの構造的な問題とは、「各列が1つの変数を表し、各行が1つの観測値を表し、各セルが1つの値を表す」という3つの基準を満たしていない状態を指す。データの内容上の問題には、欠損データ、重複データ、無効なデータなどが含まれる。

### データの整然性の評価

```python
original_data.sample(10)
```

抽出した10行のサンプルデータを見る限り、データは「各列が1つの変数を表し、各行が1つの観測値を表し、各セルが1つの値を表す」という基準を満たしている。

具体的には、各行はある商品の1回の取引を表し、各列は取引に関連する各変数を表しているため、構造的な問題は存在しない。

### データのクリーンさの評価

```python
original_data.info()
```

出力結果を見ると、データには合計541,909件の観測値があり、`Description`と`CustomerID`の変数に欠損値が存在していることが分かる。

また、`InvoiceDate`のデータ型は日付型、`CustomerID`のデータ型は文字列型であるべきであるため、データ型の変換を行う必要がある。

#### 欠損データの評価

`Description`に欠損値が存在することを確認した後、条件を指定して欠損値を含む観測値を抽出する。

```python
original_data[original_data["Description"].isnull()]
```

`Description`変数の値が欠損している取引データは1,454件存在する。

出力結果を見ると、これらの`Description`が欠損している取引データでは、`UnitPrice`がすべて0であることが分かる。この推測を検証するために、さらに抽出条件を追加し、`Description`変数が欠損しており、かつ`UnitPrice`が0ではないデータが存在するかどうかを確認する。

```python
original_data[(original_data["Description"].isnull()) & (original_data["UnitPrice"] != 0)]
```

抽出結果の件数は0件であり、`Description`の値が欠損しているデータは、同時に有効な`UnitPrice`の値も持っていないことが分かる。

`Description`は商品名を表し、`UnitPrice`は商品の単価を表す。いずれも今後の商品取引分析において重要な変数である。これらが同時に欠損または無効である場合、そのデータは有効な意味を提供できないと考えられるため、後続の処理で削除することができる。

`CustomerID`変数にも同様に欠損値が存在するため、条件を指定して欠損値を含む観測値を抽出する。

```python
original_data[original_data["CustomerID"].isnull()]
```

`CustomerID`は顧客番号を表す変数であり、売れ筋商品を分析するうえで必須の変数ではない。

また、出力結果を見ると、`CustomerID`が欠損している取引データの中にも有効なデータが存在していることが分かる。そのため、`CustomerID`が空欄の観測値はそのまま残すこととする。

#### 重複データの評価

データ変数の意味から見ると、`InvoiceNo`、`StockCode`、`CustomerID`はいずれも一意の識別子である。

しかし、1回の取引には複数の商品が含まれる可能性があるため、`InvoiceNo`は重複する場合がある。また、異なる取引に同じ商品が含まれる可能性があるため、`StockCode`も重複する場合がある。さらに、顧客は複数回取引を行ったり、複数の商品を注文したりする可能性があるため、`CustomerID`も重複する場合がある。

したがって、このデータセットについては、重複データを評価する必要はない。

#### 不一致データの評価

不一致データは`Country`変数に存在する可能性があるため、同一の国を指す複数の異なる値が存在するかどうかを確認する。

```python
original_data["Country"].value_counts()
```

`Country`変数の値を見ると、`"USA"`と`"United States"`はいずれもアメリカを表しており、`"United Kingdom"`、`"UK"`、`"U.K."`はいずれもイギリスを表している。

そのため、クリーニングの段階でこれらの値を統一し、1つの表記にそろえる必要がある。

#### 無効または誤ったデータの評価

DataFrameの`describe`メソッドを使用することで、数値データの統計情報を素早く把握することができる。

```python
original_data.describe()
```

出力結果を見ると、`Quantity`と`UnitPrice`に負の値が存在しており、後続の数値分析に影響を与える可能性がある。

そのため、まず`Quantity`の値が負数である観測値を抽出し、その意味をさらに評価する。

```python
original_data[original_data["Quantity"] < 0]
```

抽出結果を見ると、`Quantity`変数が負数である観測値では、`InvoiceNo`がいずれもCで始まっているように見える。これは、注文がキャンセルされたことを表していると考えられる。

ただし、この推測が必ずしも正しいとは限らない。そこで、この推測を検証するために、さらに抽出条件を追加し、`Quantity`変数が負数であり、かつ`InvoiceNo`がCで始まらない観測値が存在するかどうかを確認する。

```python
original_data[(original_data["Quantity"] < 0) & (original_data["InvoiceNo"].str[0] != "C")]
```

上記の推測は誤りであることが確認された。なぜなら、`Quantity`変数が負数であり、かつ`InvoiceNo`がCで始まらない観測値も存在しているためである。

ただし、上記の出力結果によると、これらの抽出された観測値の`UnitPrice`はすべて0である。そのため、`UnitPrice`の条件を追加して検証する。

```python
original_data[(original_data["Quantity"] < 0)&(original_data["InvoiceNo"].str[0] != "C")&(original_data["UnitPrice"] != 0)]
```

出力結果により、この推測は検証された。`Quantity`変数が負数である場合、観測値は以下のいずれかの条件を満たしている。

- `InvoiceNo`がCで始まっており、注文がキャンセルされたことを示している
- `UnitPrice`が0であり、単価が0ポンドであることを示している

これらの取引データはいずれも有効な成約データではなく、売上に貢献しないため、後続の分析対象には含めない。

したがって、データクリーニングの段階で、`Quantity`が負数である観測値を削除する。

次に、`UnitPrice`が負数である観測値を抽出する。

```python
original_data[original_data["UnitPrice"] < 0]
```

出力結果を見ると、`UnitPrice`が負数である観測値はいずれも貸倒損失の調整であり、実際の商品取引データには該当しない。

そのため、データクリーニングの段階で、これらの観測値も削除する。

## データのクリーニング

前の評価部分で得られた結論に基づき、実施する必要のあるデータクリーニングは以下のとおりである。

- `InvoiceDate`変数のデータ型を日時型に変換する
- `CustomerID`変数のデータ型を文字列型に変換する
- `Description`変数が欠損している観測値を削除する
- `Country`変数の値`"USA"`を`"United States"`に置き換える
- `Country`変数の値`"UK"`、`"U.K."`を`"United Kingdom"`に置き換える
- `Quantity`変数の値が負数である観測値を削除する
- `UnitPrice`変数の値が負数である観測値を削除する

クリーニング後のデータと元のデータを区別するために、新しい変数`cleaned_data`を作成し、`original_data`のコピーを代入する。

以降のクリーニング処理は、すべて`cleaned_data`に対して行う。

```python
cleaned_data = original_data.copy()
cleaned_data.head()
```

`InvoiceDate`変数のデータ型を日時型に変換する。

```python
cleaned_data["InvoiceDate"] = pd.to_datetime(cleaned_data["InvoiceDate"])
cleaned_data["InvoiceDate"]
```

`CustomerID`変数のデータ型を文字列型に変換する。

```python
cleaned_data["CustomerID"] = cleaned_data["CustomerID"].astype(str)
cleaned_data["CustomerID"]
```

`CustomerID`変数の値の末尾にある`".0"`を削除する。

```python
cleaned_data["CustomerID"] = cleaned_data["CustomerID"].str.slice(0, -2)
cleaned_data["CustomerID"]
```

`Description`変数が欠損している観測値を削除し、削除後にこの列の欠損値の合計数を確認する。

```python
cleaned_data.dropna(subset=["Description"], inplace=True)
```

```python
cleaned_data["Description"].isnull().sum()
```

`Country`変数の値`"USA"`を`"United States"`に置き換え、置き換え後に`"USA"`という値の件数を確認する。

```python
cleaned_data["Country"] = cleaned_data["Country"].replace({"USA": "United States"})
```

```python
len(cleaned_data[cleaned_data["Country"] == "USA"])
```

`Country`変数の値`"UK"`、`"U.K."`を`"United Kingdom"`に置き換え、置き換え後に`"UK"`と`"U.K."`という値の件数を確認する。

```python
cleaned_data["Country"] = cleaned_data["Country"].replace({"UK": "United Kingdom", "U.K.": "United Kingdom"})
```

```python
print(len(cleaned_data[cleaned_data["Country"] == "UK"]))
print(len(cleaned_data[cleaned_data["Country"] == "U.K."]))
```

`Quantity`変数の値が負数である観測値を削除し、削除後に`Quantity`変数の値が負数である件数を確認する。

```python
cleaned_data = cleaned_data[cleaned_data["Quantity"] >= 0]
```

```python
len(cleaned_data[cleaned_data["Quantity"] < 0])
```

`UnitPrice`変数の値が負数である観測値を削除し、削除後に`UnitPrice`変数の値が負数である件数を確認する。

```python
cleaned_data = cleaned_data[cleaned_data["UnitPrice"] >= 0]
```

```python
len(cleaned_data[cleaned_data["UnitPrice"] < 0])
```

データクリーニングが完了した後、クリーンで整然としたデータを新しいファイルに保存する。ファイル名は`e_commerce_cleaned.csv`とする。
